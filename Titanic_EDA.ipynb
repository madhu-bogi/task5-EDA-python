# Titanic Exploratory Data Analysis (EDA)
This notebook performs exploratory data analysis (EDA) on the Titanic dataset (`train.csv`, `test.csv`, `gender_submission.csv`).
- Author: Generated by ChatGPT
- Tools: Python (Pandas, Matplotlib, Seaborn)


```python
# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
# 2. Load Data
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
gender_df = pd.read_csv('gender_submission.csv')

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("Gender_submission shape:", gender_df.shape)

train_df.head()
```

```python
# 3. Basic Info & Summary
train_df.info()
train_df.describe()

print("Missing values:\n", train_df.isnull().sum())
print("Survived:\n", train_df['Survived'].value_counts())
print("Pclass:\n", train_df['Pclass'].value_counts())
print("Sex:\n", train_df['Sex'].value_counts())
```

```python
# 4. Univariate Analysis
train_df.hist(figsize=(12,8), bins=20)
plt.show()

sns.boxplot(x=train_df['Age'])
plt.show()

sns.boxplot(x=train_df['Fare'])
plt.show()
```

```python
# 5. Bivariate Analysis
sns.countplot(x='Sex', hue='Survived', data=train_df)
plt.title("Survival Count by Sex")
plt.show()

sns.countplot(x='Pclass', hue='Survived', data=train_df)
plt.title("Survival Count by Pclass")
plt.show()

sns.countplot(x='Embarked', hue='Survived', data=train_df)
plt.title("Survival Count by Embarked")
plt.show()
```

```python
# 6. Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(train_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# 7. Pairplot
sns.pairplot(train_df[['Survived', 'Age', 'Fare', 'Pclass']], hue='Survived')
plt.show()
```

```python
# 8. Advanced Insights
plt.figure(figsize=(8,6))
sns.kdeplot(train_df.loc[train_df['Survived']==1, 'Age'], shade=True, label="Survived")
sns.kdeplot(train_df.loc[train_df['Survived']==0, 'Age'], shade=True, label="Not Survived")
plt.title("Age Distribution by Survival")
plt.legend()
plt.show()

train_df['FamilySize'] = train_df['SibSp'] + train_df['Parch'] + 1
sns.countplot(x='FamilySize', hue='Survived', data=train_df)
plt.title("Survival Count by Family Size")
plt.show()
```
